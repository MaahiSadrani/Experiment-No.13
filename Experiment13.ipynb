```python
# Preprocessing on a dataset
data = {
    "Roll No": [101, 102, 103, 104, 105],
    "Name": ["Asha", "Rohan", "Meera", "Karan", "Anil"],
    "Age": [20, "-", 21, 20, "22"],
    "Marks": [85, 90, "-", 88, 76],
    "Department": ["CSE", "IT", "CSE", "cse", "E&TC"],
    "Admission_Date": ["01-06-2023", "03-06-2023", "05-06-2023", "07-06-2023", "10/06/23"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Step 2: Replace '-' with Nan tp marks missing value

df.replace("-",np.nan,inplace=True)
df
```

```python
# Convert columns to numeric where needed
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
df
```

```python
# Step 3: Checking thne missing values
print("\nMissing Values:")
print(df.isna().sum())
```

```python
# Step 4: Fill missing Age with mean
df["Age"].fillna(df["Age"].mean(),inplace=True)
df["Marks"].fillna(df["Marks"].median(),inplace=True)
df
```

```python
# Step 5: Fix inconsistent text for text formating
df["Department"]=df["Department"].str.upper()
df
```

```python
# Step 6: Convert admission date to standard datetime format

df["Admission_Date"]=pd.to_datetime(df["Admission_Date"],errors='coerece'
df
```
